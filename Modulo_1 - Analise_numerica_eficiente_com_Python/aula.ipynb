```python
import numpy as np
url = 'https://raw.githubusercontent.com/alura-cursos/numpy/dados/apples_ts.csv'
dados = np.loadtxt(url, delimiter= ',', usecols=np.arange(1,88,1))
dados
```

### Entendendo meus dados:

```python
dimensoes_dados = dados.ndim
print(f'Meus dados tem {dimensoes_dados} dimensões: Linhas e Colunas\n')

num_dados = dados.size
print(f'Meus dados tem {num_dados} elementos\n') # size = Número de linhas x Número de colunas

linhas_colunas = dados.shape
print(f'Meus dados tem {linhas_colunas[0]} linhas e {linhas_colunas[1]} colunas')
print(linhas_colunas)
```

```python
# array_equal() verifica se todos os elementos das duas arrays são exatamente iguais (mesmo tipo, valor e ordem)
print(np.array_equal(Moscow_ano3,Moscow_ano4)) 

# np.allclose() verifica em cada valor se se seus respectivos elementos são "aproximadamente iguais" dentro de uma tolerância que nesse caso é 20
print(np.allclose(Moscow_ano3,Moscow_ano4,atol=20))
```

```python
#Verificar a quantidade de nan
#A função isnan() retorna valores como falsos e verdadeiros do nosso array, usaremos sum() pois cada valor positivo é interpretado como 1
sum(np.isnan(Kaliningrad))
```

### Lidando com NaN

Iremos substituir o mesmo pela média dos dois mais próximos dele

```python
Kaliningrad[4] = np.mean([Kaliningrad[3],Kaliningrad[5]])
plt.plot(datas_totais, Kaliningrad)
```
